```python
import pandas as pd
import pickle
```

```python
instruct_res = pd.read_csv('/home/s6moakba/Instruct_Restaurants.csv')
```

```python
import ast

instruct_res['aspect_terms'] = instruct_res['aspect_terms'].apply(ast.literal_eval)
```

```python
instruct_restaurant_nouns = set().union(*instruct_res['aspect_terms'])
```

```python
with open('instruct_restaurant_nouns.pkl', 'wb') as file:
    pickle.dump(instruct_restaurant_nouns, file)
```

```python
import pickle

with open('instruct_restaurant_nouns.pkl', 'rb') as file:
    instruct_nouns = pickle.load(file)

print(len(instruct_nouns))
```
