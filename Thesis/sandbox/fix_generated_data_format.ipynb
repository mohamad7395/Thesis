# filtering based on intended aspect present in generated Text


```python
gen_data
```

```python
def filter_rows(row):
    aspects = row['aspect']
    if all(aspect in row['generated_text'] for aspect in aspects):
        row['correct'] = True    
    else:
        row['correct'] = False
    return row

gen_data = gen_data.apply(filter_rows, axis=1)            
```

```python
gen_data = gen_data[gen_data['correct'] == True]
```

```python
gen_data.shape
```

```python
gen_data = gen_data.drop(columns=['correct'])
```

```python
def create_aspect_terms(aspect,polarity):
    aspect_terms = []
    for aspect, polarity in zip(aspect, polarity):
        aspect_terms.append({'term': aspect, 'polarity': polarity})
    return aspect_terms    
    
```

```python
gen_data['aspectTerms'] = gen_data.apply(lambda x: create_aspect_terms(x['aspect'],x['polarity']),axis=1)
```
