```python
def mask_phrases(text, phrases):
    """
    For each phrase in 'phrases':
      - If the phrase has multiple words, replace each word with [MASK].
      - If the phrase has only one word, check it is a standalone match,
        then replace it with a single [MASK].
    """
    # 1) Sort phrases by descending number of words (so multi-word phrases get matched first).
    sorted_phrases = sorted(phrases, key=lambda p: len(p.split()), reverse=True)
    
    # 2) Tokenize the text into words/punctuation. 
    #    E.g., "dance floor," -> ["dance", "floor", ","]
    tokens = re.findall(r"\w+|[^\w\s]+", text)

    i = 0
    while i < len(tokens):
        matched_any = False
        
        # Try each phrase at the current token index
        for phrase in sorted_phrases:
            # Tokenize the phrase similarly
            phrase_tokens = re.findall(r"\w+|[^\w\s]+", phrase)
            phrase_len = len(phrase_tokens)
            
            # Only attempt to match if there's enough room left in 'tokens'
            if i + phrase_len <= len(tokens):
                # Check if all tokens match (case-insensitive)
                match = True
                for j in range(phrase_len):
                    if tokens[i + j].lower() != phrase_tokens[j].lower():
                        match = False
                        break

                # If we have a match, do the masking
                if match:
                    if phrase_len == 1:
                        # Special case: single word
                        # It's already a 'standalone' match if we got here, 
                        # because tokenizing ensures we won't partially match "pizza" in "pizzeria".
                        tokens[i] = "[MASK]"
                    else:
                        # Multi-word phrase: replace each token in the phrase with [MASK]
                        for j in range(phrase_len):
                            tokens[i + j] = "[MASK]"
                    
                    # Advance the pointer by the length of the matched phrase
                    i += phrase_len
                    matched_any = True
                    break  # Stop checking other phrases at this index
        
        # If no phrase matched at this index, move on
        if not matched_any:
            i += 1
    
    # Reconstruct a string from the tokens
    return " ".join(tokens)
```

```python
training_set['text_blank'] = training_set.apply(lambda row: mask_phrases(row['raw_text'], row['terms']), axis=1)
```

```python
import random
def create_prompt():
    prompt = prompt_begining
    prompt  += "\n" + random.choice(additional_instruction)
    prompt += "\n" + "-Good examples:"
    prompt += "\n" + random.choice(examples_len_1)
    prompt += "\n" + random.choice(examples_len_2)
    prompt += "\n" + random.choice(examples_len_3) 
    prompt += "\n" +"""Now complete this task like examples above: \n### Input ###"""
    return prompt
```

```python

def preprocess_data(df,iter):
    data = []
    for _, row in df.iterrows():
        for i  in range(iter):
            blanked_text = row['text_blank']
            prompt = create_prompt() + f"\nsentence: {blanked_text}\n###Output### "
            data.append({
                'blank': blanked_text,
                'original_terms': row['terms'],
                'original_text': row['raw_text'],
                'prompt': prompt,
                'aspectTerms_old': row['aspectTerms'],
            })
    return pd.DataFrame(data)
```
