# INSTRUCTABSA performance

```python
with open("/home/s6moakba/InstructABSA/label_compare_test.txt", "r") as file:
    content_ad = file.read()

lines = content_ad.split('\n')
id_te_labels_ad = eval(lines[0].split(': ')[1])
id_te_pred_labels_ad = eval(lines[1].split(': ')[1])
```

```python
correct_pred = 0
for i in range(len(id_te_labels_ad)):
    if id_te_labels_ad[i] == id_te_pred_labels_ad[i]:
        correct_pred+=1 
print('INSTRUCTABSA correct prediction percentage: ',correct_pred/len(id_te_labels_ad))
```
