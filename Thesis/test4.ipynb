```python
with open("/home/s6moakba/InstructABSA/label_compare_validation.txt", "r") as file:
    content_og = file.read()
    
lines_og = content_og.split('\n')
id_te_labels_og = eval(lines_og[0].split(': ')[1])
id_te_pred_labels_og = eval(lines_og[1].split(': ')[1])
```

```python
with open("/home/s6moakba/InstructABSA/label_compare_validation_new_data.txt", "r") as file:
    content_ad = file.read()

lines = content_ad.split('\n')
id_te_labels_ad = eval(lines[0].split(': ')[1])
id_te_pred_labels_ad = eval(lines[1].split(': ')[1])
```

```python
missed_examples_og = []

for item in zip(id_te_labels_og, id_te_pred_labels_og):
    if item[0] != item[1]:
        missed_examples_og.append(item)
```

```python
missed_examples_ad = []

for item in zip(id_te_labels_ad, id_te_pred_labels_ad):
    if item[0] != item[1]:
        missed_examples_ad.append(item)
```

```python
len_label_missed_example_og = []
for item in missed_examples_og:
    item_label = item[0].split(', ')
    item_pred_label = item[1].split(', ') 
    len_label_missed_example_og.append(len(item_label))
```

```python
len_label_missed_example_ad = []
for item in missed_examples_ad:
    item_label = item[0].split(', ')
    item_pred_label = item[1].split(', ') 
    len_label_missed_example_ad.append(len(item_label))
```

```python
len_label_all_example_og = []
for item in id_te_labels_og:
    item_label = item.split(', ')
    len_label_all_example_og.append(len(item_label))
```

```python
unique_values_all_examples = set(len_label_all_example_og)
value_counts_all_examples = {value: len_label_all_example_og.count(value) for value in unique_values_all_examples}

unique_values_missed_example_og = set(len_label_missed_example_og)
value_counts_missed_example_og = {value: len_label_missed_example_og.count(value) for value in unique_values_missed_example_og}


unique_values_missed_example_ad = set(len_label_missed_example_ad)
value_counts_missed_example_ad = {value: len_label_missed_example_ad.count(value) for value in unique_values_missed_example_ad}

for value, count in value_counts_all_examples.items():
    print(f"{value}: {count}")
    print(f"Count in all examples: {value_counts_all_examples.get(value, 0)}")
    print(f"Count in missed examples original: {value_counts_missed_example_og.get(value, 0)}")
    print(f"Count in missed examples new data: {value_counts_missed_example_ad.get(value, 0)}")
    ratio_og = value_counts_missed_example_og.get(value, 0) / value_counts_all_examples.get(value, 1)
    ratio_ad = value_counts_missed_example_ad.get(value, 0) / value_counts_all_examples.get(value, 1)
    print(f"Ratio before data added: {ratio_og}")
    print(f"Ratio after data added: {ratio_ad}")
```
